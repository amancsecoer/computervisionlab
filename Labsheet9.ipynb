**Check whether a number is divisible by 2, 3, or both**

```python
num = int(input("Enter a number: "))

if num % 2 == 0 and num % 3 == 0:
    print("The number is divisible by both 2 and 3.")
elif num % 2 == 0:
    print("The number is divisible by 2.")
elif num % 3 == 0:
    print("The number is divisible by 3.")
else:
    print("The number is not divisible by 2 or 3.")
```

**Check whether a year is a leap year or not**

```python
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```

**Check whether a given character is a vowel, consonant, or neither**

```python
char = input("Enter a single character: ")

if len(char) == 1 and char.isalpha():
    if char.lower() in 'aeiou':
        print("It is a vowel.")
    else:
        print("It is a consonant.")
else:
    print("It is neither a vowel nor a consonant.")
```

**Accept three sides of a triangle and determine if it is valid and classify it**

```python
a = float(input("Enter side a: "))
b = float(input("Enter side b: "))
c = float(input("Enter side c: "))

# Check if the triangle is valid
if a + b > c and a + c > b and b + c > a:
    print("The triangle is valid.")

    if a == b == c:
        print("It is an equilateral triangle.")
    elif a == b or b == c or a == c:
        print("It is an isosceles triangle.")
    else:
        print("It is a scalene triangle.")
else:
    print("The triangle is not valid.")
```

**Calculate the electricity bill based on units consumed**


```python
units = int(input("Enter the number of units consumed: "))

if units <= 100:
    bill = units * 1
elif units <= 200:
    bill = 100 * 1 + (units - 100) * 2
else:
    bill = 100 * 1 + 100 * 2 + (units - 200) * 3

print(f"Total electricity bill: ₹{bill}")
```

**Check if a given number is a palindrome**

```python
num = input("Enter a number: ")

if num == num[::-1]:
    print(f"{num} is a palindrome.")
else:
    print(f"{num} is not a palindrome.")
```

** Calculate the grade of a student based on marks**

```python
marks = float(input("Enter your marks: "))

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

**Check voting eligibility based on age and nationality**

```python
age = int(input("Enter your age: "))
nationality = input("Enter your nationality: ")

if age >= 18 and nationality.lower() == "indian":
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")
```

**Check if a number is positive, negative, or zero**

```python
num = float(input("Enter a number: "))

if num > 0:
    print("The number is positive.")
elif num < 0:
    print("The number is negative.")
else:
    print("The number is zero.")
```
