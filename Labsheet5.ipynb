```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
import ipywidgets as widgets

from IPython.display import display

print("Libraries imported successfully!")
```

```python
from google.colab import files

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("File:", file_name)
```

```python
print("First 5 rows of the dataset:")

display(df.head())
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
print("Missing Values in Each Column:")

print(df.isnull().sum())
```

```python
print("Rows before removing duplicates:", len(df))

df = df.drop_duplicates()

print("Rows after removing duplicates:", len(df))
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

print("Date columns converted successfully!")

print(df[['Order Date', 'Ship Date']].head())
```

```python
print("Statistical Summary:")

display(df.describe())
```

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_orders = df['Order ID'].nunique()
average_order_value = total_sales / total_orders
customer_count = df['Customer ID'].nunique()

print("======================================")
print("           BUSINESS KPIs")
print("======================================")

print(f"Total Sales         : ${total_sales:,.2f}")
print(f"Total Profit        : ${total_profit:,.2f}")
print(f"Total Orders        : {total_orders:,}")
print(f"Average Order Value : ${average_order_value:,.2f}")
print(f"Customer Count      : {customer_count:,}")

print("======================================")
```

```python
fig = go.Figure()

fig.add_trace(go.Indicator(
    mode="number",
    value=total_sales,
    title={"text": "Total Sales"},
    number={"prefix": "$", "valueformat": ",.0f"},
    domain={'x': [0, 0.2], 'y': [0, 1]}
))

fig.add_trace(go.Indicator(
    mode="number",
    value=total_profit,
    title={"text": "Total Profit"},
    number={"prefix": "$", "valueformat": ",.0f"},
    domain={'x': [0.2, 0.4], 'y': [0, 1]}
))

fig.add_trace(go.Indicator(
    mode="number",
    value=total_orders,
    title={"text": "Total Orders"},
    number={"valueformat": ","},
    domain={'x': [0.4, 0.6], 'y': [0, 1]}
))

fig.add_trace(go.Indicator(
    mode="number",
    value=average_order_value,
    title={"text": "Average Order Value"},
    number={"prefix": "$", "valueformat": ",.2f"},
    domain={'x': [0.6, 0.8], 'y': [0, 1]}
))

fig.add_trace(go.Indicator(
    mode="number",
    value=customer_count,
    title={"text": "Customers"},
    number={"valueformat": ","},
    domain={'x': [0.8, 1], 'y': [0, 1]}
))

fig.update_layout(
    title="Business Performance KPIs",
    height=300
)

fig.show()
```

```python
category_sales = (
    df.groupby('Category')['Sales']
    .sum()
    .reset_index()
    .sort_values('Sales', ascending=False)
)

fig = px.bar(
    category_sales,
    x='Category',
    y='Sales',
    title='Total Sales by Product Category',
    text_auto='.2s'
)

fig.show()
```

```python
category_profit = (
    df.groupby('Category')['Profit']
    .sum()
    .reset_index()
    .sort_values('Profit', ascending=False)
)

fig = px.bar(
    category_profit,
    x='Category',
    y='Profit',
    title='Total Profit by Product Category',
    text_auto='.2s'
)

fig.show()
```

```python
monthly_sales = (
    df.groupby(df['Order Date'].dt.to_period('M'))['Sales']
    .sum()
    .reset_index()
)

monthly_sales['Order Date'] = (
    monthly_sales['Order Date'].astype(str)
)

fig = px.line(
    monthly_sales,
    x='Order Date',
    y='Sales',
    title='Monthly Sales Trend',
    markers=True
)

fig.update_layout(
    xaxis_title='Month',
    yaxis_title='Sales'
)

fig.show()
```

```python
region_sales = (
    df.groupby('Region')['Sales']
    .sum()
    .reset_index()
    .sort_values('Sales', ascending=False)
)

fig = px.bar(
    region_sales,
    x='Region',
    y='Sales',
    title='Total Sales by Region',
    text_auto='.2s'
)

fig.show()
```

```python
region_profit = (
    df.groupby('Region')['Profit']
    .sum()
    .reset_index()
    .sort_values('Profit', ascending=False)
)

fig = px.bar(
    region_profit,
    x='Region',
    y='Profit',
    title='Total Profit by Region',
    text_auto='.2s'
)

fig.show()
```

```python
segment_sales = (
    df.groupby('Segment')['Sales']
    .sum()
    .reset_index()
)

fig = px.pie(
    segment_sales,
    names='Segment',
    values='Sales',
    title='Sales Distribution by Customer Segment'
)

fig.show()
```

```python
top_products = (
    df.groupby('Product Name')['Sales']
    .sum()
    .reset_index()
    .sort_values('Sales', ascending=False)
    .head(10)
)

fig = px.bar(
    top_products.sort_values('Sales'),
    x='Sales',
    y='Product Name',
    orientation='h',
    title='Top 10 Products by Sales',
    text_auto='.2s'
)

fig.show()
```

```python
subcategory_sales = (
    df.groupby('Sub-Category')['Sales']
    .sum()
    .reset_index()
    .sort_values('Sales', ascending=False)
)

fig = px.bar(
    subcategory_sales,
    x='Sub-Category',
    y='Sales',
    title='Sales by Sub-Category',
    text_auto='.2s'
)

fig.show()
```

```python
fig = px.scatter(
    df,
    x='Discount',
    y='Profit',
    size='Sales',
    color='Category',
    hover_data=['Product Name', 'Region'],
    title='Relationship Between Discount and Profit'
)

fig.show()
```

```python
categories = ['All'] + sorted(df['Category'].unique())

category_dropdown = widgets.Dropdown(
    options=categories,
    description='Category:',
    style={'description_width': 'initial'}
)

output = widgets.Output()

def category_dashboard(change=None):
    with output:
        output.clear_output(wait=True)

        selected_category = category_dropdown.value

        if selected_category == 'All':
            filtered_df = df.copy()
        else:
            filtered_df = df[
                df['Category'] == selected_category
            ]

        sales = filtered_df['Sales'].sum()
        profit = filtered_df['Profit'].sum()
        orders = filtered_df['Order ID'].nunique()

        print("====================================")
        print("       FILTERED DASHBOARD")
        print("====================================")
        print(f"Category      : {selected_category}")
        print(f"Sales         : ${sales:,.2f}")
        print(f"Profit        : ${profit:,.2f}")
        print(f"Orders        : {orders:,}")

        region_data = (
            filtered_df.groupby('Region')['Sales']
            .sum()
            .reset_index()
        )

        fig = px.bar(
            region_data,
            x='Region',
            y='Sales',
            title=f'Sales by Region — {selected_category}',
            text_auto='.2s'
        )

        fig.show()

category_dropdown.observe(
    category_dashboard,
    names='value'
)

display(category_dropdown)
display(output)

category_dashboard()
```

```python
regions = ['All'] + sorted(df['Region'].unique())

region_dropdown = widgets.Dropdown(
    options=regions,
    description='Region:',
    style={'description_width': 'initial'}
)

output_region = widgets.Output()

def region_dashboard(change=None):
    with output_region:
        output_region.clear_output(wait=True)

        selected_region = region_dropdown.value

        if selected_region == 'All':
            filtered_df = df.copy()
        else:
            filtered_df = df[
                df['Region'] == selected_region
            ]

        sales = filtered_df['Sales'].sum()
        profit = filtered_df['Profit'].sum()
        orders = filtered_df['Order ID'].nunique()

        print("====================================")
        print("       REGION DASHBOARD")
        print("====================================")
        print(f"Region   : {selected_region}")
        print(f"Sales    : ${sales:,.2f}")
        print(f"Profit   : ${profit:,.2f}")
        print(f"Orders   : {orders:,}")

        category_data = (
            filtered_df.groupby('Category')['Sales']
            .sum()
            .reset_index()
        )

        fig = px.bar(
            category_data,
            x='Category',
            y='Sales',
            title=f'Sales by Category — {selected_region}',
            text_auto='.2s'
        )

        fig.show()

region_dropdown.observe(
    region_dashboard,
    names='value'
)

display(region_dropdown)
display(output_region)

region_dashboard()
```

```python
category_filter = widgets.Dropdown(
    options=['All'] + sorted(df['Category'].unique()),
    description='Category:'
)

region_filter = widgets.Dropdown(
    options=['All'] + sorted(df['Region'].unique()),
    description='Region:'
)

segment_filter = widgets.Dropdown(
    options=['All'] + sorted(df['Segment'].unique()),
    description='Segment:'
)

dashboard_output = widgets.Output()

def update_full_dashboard(change=None):
    with dashboard_output:
        dashboard_output.clear_output(wait=True)

        filtered = df.copy()

        if category_filter.value != 'All':
            filtered = filtered[
                filtered['Category'] == category_filter.value
            ]

        if region_filter.value != 'All':
            filtered = filtered[
                filtered['Region'] == region_filter.value
            ]

        if segment_filter.value != 'All':
            filtered = filtered[
                filtered['Segment'] == segment_filter.value
            ]

        sales = filtered['Sales'].sum()
        profit = filtered['Profit'].sum()
        orders = filtered['Order ID'].nunique()

        print("==========================================")
        print("       INTERACTIVE BUSINESS DASHBOARD")
        print("==========================================")
        print(f"Sales  : ${sales:,.2f}")
        print(f"Profit : ${profit:,.2f}")
        print(f"Orders : {orders:,}")

        chart_data = (
            filtered.groupby('Category')['Sales']
            .sum()
            .reset_index()
        )

        fig = px.bar(
            chart_data,
            x='Category',
            y='Sales',
            title='Filtered Sales by Category',
            text_auto='.2s'
        )

        fig.show()

category_filter.observe(
    update_full_dashboard,
    names='value'
)

region_filter.observe(
    update_full_dashboard,
    names='value'
)

segment_filter.observe(
    update_full_dashboard,
    names='value'
)

display(category_filter)
display(region_filter)
display(segment_filter)
display(dashboard_output)

update_full_dashboard()
```
